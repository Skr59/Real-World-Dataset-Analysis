### Problem Statement & Business Context
In modern multi-region e-commerce marketplaces, operational efficiency—measured through delivery speed, logistics accuracy, and customer satisfaction—directly dictates customer retention and seller success. 

The goal of this analysis is to evaluate order fulfillment cycles, delivery delay dynamics, freight cost structures, and review score distributions to uncover operational bottlenecks and actionable business insights.

### Primary Analytical Questions
1. How accurately do estimated delivery dates match actual customer delivery times across different regions?
2. What are the key drivers (freight cost, delivery delay, product category) impacting customer review scores?
3. How do sales volumes and logistics delays vary geographically across Brazilian states?

### Key Dataset Variables
- `order_id` / `customer_id`: Unique order and customer identifiers.
- `order_status`: Fulfillment stage (e.g., delivered, canceled, shipped).
- `order_purchase_timestamp`: Datetime when the order was placed.
- `order_delivered_customer_date`: Actual delivery date to the customer.
- `order_estimated_delivery_date`: Estimated delivery date promised at checkout.
- `price` / `freight_value`: Item purchase price and shipping logistics charge.
- `review_score`: Customer review score ranging from 1 to 5.
