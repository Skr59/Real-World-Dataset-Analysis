# E-commerce Analysis

This notebook loads the e-commerce orders dataset and prepares the date fields for analysis.

```python
import os
import glob

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

candidate_paths = [
    "data/olist_orders_dataset.csv",
    "data/raw/olist_orders_dataset.csv",
    "olist_orders_dataset.csv",
    "data/orders.csv",
    "data/raw/orders.csv",
    "orders.csv",
]

csv_path = next((path for path in candidate_paths if os.path.exists(path)), None)
if csv_path is None:
    matches = sorted(glob.glob("**/*orders*.csv", recursive=True))
    if matches:
        csv_path = matches[0]
    else:
        raise FileNotFoundError("No orders CSV found. Add the dataset to data/ or data/raw/ before running the notebook.")

df_orders = pd.read_csv(csv_path)

datetime_cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
]
for col in datetime_cols:
    if col in df_orders.columns:
        df_orders[col] = pd.to_datetime(df_orders[col], errors='coerce')

df_orders.head()
```
